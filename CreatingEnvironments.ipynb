# Python Package Versioning, Conflicts, and Dependencies

As you now know, Python is comprised of a core language and additional packages we can import (e.g. Pandas, NumPy, math, matplotlib) that add functionality to the core language. The Python core language and all Python packages release new versions with varying frequency. For example, take a look at the timing of version releases below for core Python, Pandas, and Xarray. Major new version releases of core Python occur about once per year, Pandas releases updated versions approximately every 1-6 months and Xarray releases updated versions approximately monthly. 

![release schedules for python pandas and xarray](images/envs_ReleaseSchedules.png)

New versions may include new features, bug fixes, security updates, and more. With different packages releasing new versions on different schedules and, in many cases, more often than the time it takes to write complete code for any given research project, keeping up with the latest version of all the packages you use is nearly impossible. Also, sometimes new package versions will *break* code you wrote that worked with older package versions, meaning that code you wrote a year ago may throw errors if you try to run it using a updated package versions. Additionally, since packages are developed by different entities or institutions, not all packages or versions of packages are compatible with each other. When this happens it's called a *package conflict*. And finally, each package may require the installation of other packages called *dependencies* to work. You're probably thinking that it sounds very complicated to know and manage all the information about new package releases, package conflicts, and dependencies, and you're right! Don't worry though, this is where Anaconda and Python environments make our lives so much easier. When you create an environment with Anaconda, Anaconda will take care of all of this for you.

# Anaconda for Package Installation and Environment Management

For this course, we have chosen to use Anaconda for installing Python (core language and packages) and creating environments because it is the most popular choice for geosciences. There are other ways to install Python and create environments, but we won't cover them. When you installed Anaconda, a specific version of core Python and specific versions of frequently used Python packages were installed to your computer. All of the packages we have imported in our code thus far were already installed with Anaconda, so to access them in our code we simply needed to use an import statement. But there are thousands of Python packages (including all the different versions of each package) and Anaconda doesn't install them all because that would take up way too much space on your computer. What would we do if we need packages that Anaconda didn't install automatically? This is the situation we'll encounter in the next lessons on GeoPandas and Xarray. The solution is to create a Python environment with Anaconda where we can install all the packages we need and specify particular versions of packages, if needed. 

An *environment* is a directory on your computer that contains a specific collection of packages that you have installed. We can point to an environment we've created on our computer to run Python code, which will execute using only the packages we've installed to that environment. In this course we'll create *conda environments* which are just Python environments created with Anaconda. 

**Here's why conda environments are useful:**
- **We can create many different environments, all of which will be separate from each other.** This is useful if we need different Python packages for different projects. It's recommended to create a new conda environment for each new project or workflow.  
- **With different environments, we can keep track of the packages and package versions that are needed to run any given code.** For example, we might have old code we wrote 5 years ago that needs older versions of some packages to run (i.e., updated package versions break the old code). We can ensure that we'll be able to run our code without it breaking if we create an environment with the older package versions and then always point to that environment when we want to run our old code.
- **We can export the details of what's inside any conda environment and save it to a special file such that we can delete and recreate the environment at a later date or share the environment with collaborators.** You'll see this file referred to by many names such as yml file, yaml file, spec/specification file, requirements file, export file, environment file, and probably more. We can name these special files anything we want but the file extension is always .yml or .yaml. This feature of environments is nice for reproducibility and collaboration.

# Building environments with Anaconda

Let's walk through the process of *building* (creating) a conda environment before we actually attempt it.

1) Use the Anaconda Prompt (or terminal on Mac/Linux) to name our new environment and request that certain packages be installed to it.
2) Anaconda will then search through huge online repositories of packages to find the packages we asked for and all the necessary dependencies. Anaconda will attempt to get the most up-to-date version of everything we asked for unless we asked for specific versions of packages. 
3) Next, Anaconda will determine if there are any conflicts between all the packages and dependencies. If there are conflicts, Anaconda will solve the problem by adusting package versions until it finds a combination of packages and dependencies that are all compatible with each other.
4) Anaconda will then download the final combination of packages and dependencies to our environment.
5) Finally, Anaconda will install packages and double check that everything installed correctly.

When the process is complete, Anaconda will say that it's done and the terminal prompt where you type commands will reappear. This whole process can take anywhere from 30 seconds to 10 minutes (or possibly more) depending on the size of the packages we request, if there are a lot of package conflicts to solve, and the speed of your internet connection. So be patient and don't give up prematurely!

## Build your first conda environment

Now we can move on to building a new environment! We'll build a small test environment first. 

Using the Anaconda Prompt (Window's users) or terminal (Max/Linux users) you're going to issue the following command. You may want to read through this info and watch the short video before you do this yourself.

```
conda create --name test --channel conda-forge geopandas ipykernel
```

**What's going on here? Let's break it down.**
- ```conda create``` is the command to create a new environment.
- ```--name test``` envokes the option to name the new environment. We're naming it ```test```. You always want to use this option and give each new environment you create a unique name.
- ```--channel conda-forge``` tells Anaconda that we want to search the online repository called conda forge for packages. This is the repository that is used most frequently in the geosciences. 
- ```geopandas ipykernel``` are the two packages we are requesting to be downloaded and installed to our test environment

**What do these packages do?**
- ```geopandas``` is a package that enables spatial operations on geometric types of data and we'll cover it in the next lesson.
- ```ipykernel``` is a package that turns conda environments into Jupyter kernels. We won't get into the details of Jupyter kernels in this course. All you need to know is that you need Jupyter kernels to access your conda environments inside a Jupyter notebook. **If you are planning to use Jupyter, put this package in every environment that you create.** This package in combination with nb_conda_kernels in the base environment will automatically create a Jupyter kernel for each of your conda environments and make them all accessible inside Jupyter. If you are confused, don't worry. We will look briefly at Jupyter kernels later in the lesson.

## Listing packages in an environment

When we created the test environment, we did not specify what version of geopandas or ipykernel to install. We can see what version Anaconda installed in the test environment by activating the environment and then listing information for the geopandas package. The two commands are:

```conda activate test```

```conda list geopandas```

Notice how after the first command the environment name at the beginning of the prompt changes to ```(test)```, indicating that the active environment has changed from base to test. The second command will then list any packages in the active environment whose name contains geopandas. At the time the instructor wrote this lesson, Anaconda installed geopandas version 1.0.1.
