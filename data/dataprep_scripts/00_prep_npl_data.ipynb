```python
import geopandas as gpd
import pyogrio
```

```python
data_dir=r'C://Users/kerrie/00_code/repos/DEV_canvas_beginner_py/data/EPA_superfund_boundaries/'
orig_dir = 'NPL_Boundaries.gdb'
```

# load and subset

```python
# get layer names
pyogrio.list_layers(data_dir+orig_dir)
```

```python
# preview layer data in a geodataframe
gdf = gpd.read_file(data_dir+orig_dir, layer = 'SITE_BOUNDARIES_SF')
gdf.head()
```

```python
# see all column names
gdf.columns.tolist()
```

```python
# subset and prep for writing to disk

# subset columns 
gdf = gpd.read_file(data_dir+orig_dir, layer = 'SITE_BOUNDARIES_SF',
                      columns = ['SITE_NAME','URL_ALIAS_TXT','NPL_STATUS_CODE','COUNTY','STATE_CODE'])

# subset to Mississippi
gdf = gdf.loc[gdf.STATE_CODE.str.contains('MS')].reset_index(drop=True)

# make column names l.e. 10 characters to avoid truncation when writing to disk
gdf.rename(columns={'NPL_STATUS_CODE':'NPL_STATUS','STATE_CODE':'STATE','URL_ALIAS_TXT':'URL'}, inplace=True)
gdf
```

# write to new geodatabase

```python
gdf.to_file(data_dir+'MS_EPA_NPL_Site_Boundaries.gdb')
```
