```python
import geopandas as gpd
import pyogrio
```

```python
data_dir=r'C://Users/kerrie/01_data/boundaries/states_2023/'
orig_dir = 'orig/cb_2023_us_state_500k/'
shapefile='cb_2023_us_state_500k.shp'

states_list = ['FL','GA','SC','NC'] 
```

# load and subset county shapefile

```python
states=gpd.read_file(data_dir+orig_dir+shapefile,
                      columns = ['STATEFP','STUSPS','NAME'])

print(states.crs)
states
```

```python
states.dtypes
```

```python
states = states.loc[states.STUSPS.isin(states_list)].reset_index(drop=True)
states
```

# write to new shapefile

```python
states.to_file(data_dir+'FL_GA_SC_NC/FL_GA_SC_NC_cb_2023_us_state_500k.shp')
```
