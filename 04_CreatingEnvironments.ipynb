# Python Package Versioning, Conflicts, and Dependencies

As you now know, Python is comprised of a core language and additional packages that we can import (e.g., Pandas, NumPy, math, matplotlib) to add functionality to the core language. The Python core language (which is a package itself) and all other Python packages release new versions with varying frequency. For example, take a look at the timing of version releases below for core Python, Pandas, and Xarray. Major new version releases of core Python occur about once per year, Pandas releases updated versions approximately every 1-6 months, and Xarray releases updated versions approximately monthly. 

![release schedules for python pandas and xarray](images/Envs_ReleaseSchedules.png)

New versions may include new features, bug fixes, security updates, and more. With different packages releasing new versions on different schedules and, in many cases, more often than the time it takes to write a complete code for any given research project, keeping up with the latest version of all the packages you use is nearly impossible since Python is always evolving. Also, sometimes new package versions will *break* code you wrote that worked with older package versions. This means that code you wrote a year ago may throw errors if you try to run it using updated package versions. Additionally, since packages are developed by different entities or institutions, not all packages or versions of packages are compatible with each other. When this happens it's called a *package conflict*. And finally, each package may require the installation of other packages called *dependencies* to work. 

You're probably thinking that it sounds very complicated to know and manage all the information about new package releases, package conflicts, and dependencies, and you're right! Don't worry though, this is where Anaconda and Python environments make our lives so much easier. When you create an environment with Anaconda, Anaconda can take care of all of this for you.

# Anaconda for Package Installation and Environment Management

For this course, we have chosen to use Anaconda for installing Python (core language and additional 
packages) and creating environments because it is the most popular choice for geosciences. There are other ways to install Python and create environments, but we won't cover them. When you installed Anaconda, a specific version of core Python and specific versions of frequently used Python packages were installed to your computer. All of the packages we have imported in our code thus far were already installed with Anaconda, so to access them in our code we simply needed to use an import statement. But there are thousands of Python packages (including all the different versions of each package) and Anaconda doesn't install them all because that would take up way too much space on your computer. What would we do if we need packages that Anaconda didn't install automatically? This is the situation we'll encounter in the next lessons on GeoPandas and Xarray. The solution is to create a Python environment with Anaconda where we can install all the packages we need and specify particular versions of packages, if needed. 

An *environment* is a directory on your computer that contains a specific collection of packages that you have installed. We can point to an environment we've created on our computer to run Python code, which will execute using only the packages we've installed to that specific environment. In this course we'll create *conda environments* which are just Python software environments created with Anaconda. 

**Here's why conda environments are useful:**
- **We can create many different environments, all of which will be separate from each other.** This is useful if we need different Python packages for different projects. It's recommended to create a new conda environment for each new project or workflow.  
- **With different environments, we can keep track of the packages and package versions that are needed to run any given code.** For example, we might have old code we wrote 5 years ago that needs older versions of some packages to run (i.e., updated package versions break the old code). We can ensure that we'll be able to run our code without it breaking if we create an environment with the older package versions and then always point to that environment when we want to run our old code.
- **We can export the details of what's inside any conda environment and save it to a special file such that we can delete and recreate the environment at a later date or share the environment with collaborators.** You'll see this file referred to by many names such as yml file, yaml file, spec/specification file, requirements file, export file, environment file, and probably more. This feature of environments is nice for reproducibility and collaboration.

# Building environments with Anaconda

Let's walk through the process of *building* (creating) a conda environment before we actually attempt it.

1) Use the Anaconda Prompt (or terminal on Mac/Linux) to name our new environment and request that certain packages be installed to it.
2) Anaconda will then search through huge online repositories of packages to find the packages we asked for and all the necessary dependencies. Anaconda will attempt to get the most up-to-date version of everything we asked for unless we asked for specific versions of packages. 
3) Next, Anaconda will determine if there are any conflicts between all the packages and dependencies. If there are conflicts, Anaconda will solve the problem by adusting package versions until it finds a combination of packages and dependencies that are all compatible with each other.
4) Anaconda will then download the final combination of packages and dependencies to our environment.
5) Finally, Anaconda will install packages and double check that everything installed correctly.
6) When the process is complete, Anaconda will print a message about activation (when building an environment) or simply a "done" message (if only installing packages to an existing environment) and the terminal prompt where you type commands will reappear.
  
This whole process can take anywhere from 30 seconds to 10 minutes (or possibly more) depending on the size of the packages we request, if there are a lot of package conflicts to solve, and the internet connection speed. So be patient and don't give up prematurely!

# Creating a record of your conda environment "recipes"

**Feel free to follow along with the commands in this section to generate your own specification files, but you don't have to.**

With reproducibility and collaboration in mind, it is often useful to save a record of what packages are in each of our conda environments. Anaconda allows us to export this information for each conda environment and save it in a file that can then be used to re-create that specific conda environment. There are a few different ways we can create these files and the methods differ mainly in the levels of detail we choose to put inside the file. It's those details that determine where we'll be able to re-create the environment. In this section, we'll cover three different approaches to creating a specification file (with file extensions .txt and .yml) for our test environment that include differing levels of detail and discuss what each level of detail is appropriate for. 

## Approach 1: A specification file for re-creating an identical environment on the same operating system, ensuring your code never breaks due to package updates

This type of file is appropriate for documenting every detail of each package that exists in a conda environment, including name, version, build string, and repository source of each package as well as the anaconda software version and operating system the environment was created with. **It's a good idea to create this file for every environment that you make so you have a record of exactly what is inside each environment and can rebuild the environment at any time, ensuring that your code will not break (on the same computer, that is).** The build string refers to the different "builds" of packages that are available from a repository. Different package builds are required for the packages to work on different operating systems (i.e. Windows, Mac, Linux) and for compatibility with different dependencies. In other words, build strings are specific to the operating system where you have built your conda environment. For this reason, if build strings are included in a specification file, re-creating the environment from that file will only work on a computer with the same operating system. So, while this method of creating a specification file is the most detailed recipe for a conda environment and it's a good idea to create a file like this for each environment for your records, this kind of specification file (containing build strings) will not work for re-creating your environment on computers with different operating systems.

The following command will save this type of specification file for our test environment to test_specification.txt at the location that is shown on your prompt (e.g., C:\Users\kerrie). The command creates a text file with a list of urls that point directly to the online location of each specific package build that is in the active environment. Feel free to activate the test environment if you haven't already (```conda activate test```) and try the command below yourself. After you execute the command, you can see what is inside the test_specification.txt file by using Jupyter's file browser to navigate to the file and open it. 

```conda list --explicit > test_specification.txt```

<img src="images/Envs_spectxt.png" width="400"/>

By the way, you can move this or any other specification file to any location on your computer. It's not the file location that matters, it's the contents. For the remainder of this lesson though, it will be assumed that your specification files are written to C:\Users\username.


## Approach 2: A specification file that will work across platforms (different operating systems), but new package versions may eventually become a problem

If we want to be able to share our conda environment with collaborators who may be using a different operating system, or maybe we, ourselves, have multiple computers with different operating systems, we need less details in our specification file. **This method is appropriate for getting yourself or others up and running quickly with a similar conda environment on different operating systems. This is not a great solution for consistently building, over time, a conda environment that will never break your code.** This is because we need to create a specification file without package build information and without all the dependencies (which are often platform specific), so this method, over time, may generate conda environments that include updated packages that could break older code. 

A cross-platform specification file can be created in .yml format with a conda command but it will also require manual edits of the contents. A yml file is a type of specification file where the contents follow a few formatting rules. The following command, with the ```--from-history``` option will create the "bare bones" file test_shareable.yml at the location that is shown on your prompt (e.g., C:\Users\kerrie). The ```--from-history``` option tells Anaconda to only include information about the packages that were specifically requested during the ```conda create``` step of building the test environment (in this case, only geopandas and ipykernel). This file, after some manual editing, will enable yourself or others to re-create the test conda environment on different operating systems. Feel free to activate the test environment if you haven't already (```conda activate test```) and try the command below yourself. After you execute the command, you can see what is inside the test_shareable.yml file by using Jupyter's file browser to navigate to the file and open it. 

```conda env export --from-history > test_shareable.yml```

<img src="images/Envs_shareyml.png" width="300"/>

Unfortunately, Anaconda doesn't currently offer a single command that can generate a file that will work across operating systems without some manual edits. The next step is to manually edit the file to 1) include the correct repository channels (i.e., conda forge) and 2) to remove the prefix line that specifies the location of the environment on your specific computer. After manual edits, the file should include the conda-forge channel first followed by the defaults channel in the channels section, and no prefix line.

<img src="images/Envs_ShareYmlAfterEdits.png" width="300"/>

Note that, while this method will work for creating a conda environment on computers with different operating systems, the environment when created on any one operating system will be a little different from the environment when created on another operating system. This is because different operating systems require different dependencies. The environments across platforms will be very similar, but it is possible that Anaconda could choose different versions of packages when re-creating the environment on Windows vs Mac vs Linux, and this in turn, could generate errors when you run code in the environment if your code requires specific package versions. Also note, that because we don't include any package versions, over time, this file will generate an environment that could break older code. **The bottom line here is that while this method may work in the short term, it isn't a foolproof method for creating environments across platforms in the long term. The foolproof solution is to create a container using a tool such as Docker, but that is beyond the scope of this course.**

## Approach 3: A specification file that includes package versions, ensuring package updates won't break your code, likely will work cross-platform

The first method, where we generated the file test_specification.txt, would ensure future package updates won't break your code because each url in that file includes a specific package version. But, there's no chance that file could re-create the environment across platforms. This next approach is a kind of middle-of-the-road approach with respect to level of detail in the specification file. The yml file we'll generate next will include specific package versions, will not include build strings, and will not include most dependencies. **This approach creates a specification file that will generate an environment that should not break older code and there's a strong possibility that it will work cross-platform as well.** As in the previous method, we'll need to issue a conda command and then manually edit the resulting yml file. First, we'll export the contents of the environment with package versions but not build strings to the file test.yml.

```conda env export --no-builds >test.yml```

<img src="images/Envs_YmlWithVersions.png" width="400"/>

Now, we'll delete the computer-specific prefix line at the end of the file and we'll delete everything in the dependencies section EXCEPT: geopandas, ipykernel, numpy, pandas, proj, pyogrio, python, scikit-learn, scipy, shapely. 

How do we know what to delete and what not to delete? This is a bit of an art, so there's not a great explanation here. Generally, we want to keep only the packages that are major packages (i.e., we find ourselves specifying them by name often when we create conda environments) and packages that are known to cause conflicts (e.g., proj). The more things we keep, the less likely our environment is to build correctly on different operating systems. But beyond those couple of wishy-washy guidelines, the decisions about what to keep will be based on your personal experience over time with building functioning environments for your own work. 

After all the deletions, test.yml should look like the following. Note that you can choose to clean up the channels section or not, since the hard coded channels (the ones that start with https) are redundant because they are also included in "defaults". The only thing we need to ensure is that conda-forge appears first and defaults appears second.

## Recommendations for Reproducibility and Collaboration with Python and Anaconda

1) Create a new conda environment for each coding project that you work on.
2) Create at least one specification file for each one of your conda environments.
   -  At minimum, always use approach 1 for every project you work on so that you have a full record of what software you used to run your code. This will save you many headaches when you inevitably get into the situation where your old environment got deleted or corrupted and you are trying to re-create the environment to run old code.
   -  Highly recommended, also use approach 3. This approach has the potential to save many headaches too, for example, if you ever need to run your codes on a new computer.
4) Create your specification files once you've installed all the packages you need to your environment and the code you're running in that environment is working. If you end up needing to install more packages as you develop your code further, you'll need to re-create your specification files. It's also a good idea to create specification files if you are not finished developing your code but you plan on taking a break from code development. Having a record of how to re-create your environment will be helpful if something unexpected happens like your environment gets corrupted or deleted.
5) Save your specification files in a location where you'll be able to find them and easily identify which coding project they are associated with. For example, you may have a separate directory for each project you work on where you save your code and other documents. Save your specification file(s) there too. If you are a github user, save specification file(s) to each of your repos.
6) Choose a file naming convention for your specification files and stick to it.
   - It's a good idea that the environment name appear somewhere in the specification file name.
   - You may want to include an indication in the file name of what the specification file is appropriate for (e.g., is it the detailed computer-specific specification or is it shareable/cross-platform)
   - This is a personal choice, but the instructor always names specification files created with approach 1 as envname_spec.txt, files created with approach 2 as envname_shareable.yml, and files created with approach 3 as envname.yml. You can choose whatever method you want for naming your specification file(s), just be consistent.

# Accessing conda environments inside Jupyter with Jupyter Kernels

As mentioned earlier, a Jupyter kernel is what allows us to access our conda environments inside a Jupyter notebook. We've set ourselves up such that Jupyter and Anaconda will work together to automatically create a Jupyter kernel for each conda environment on our computer. # Fixing the CondaSSLError

**Only follow the instructions in this section if you have encountered the specific error:** 

```CondaSSLError: Encountered an SSL error. Most likely a certificate verification issue.```

You may get this error when trying to use Anaconda due to the way your institution has configured their network or computers. 

**Solution:**

At your prompt, type the following command, then try again to execute whatever command you received the error from. 

```conda config --set ssl_verify false```
